# **Introducción: ¿qué es la interpretabilidad?**

<div style="height:4px;border-radius:2px;margin:4px 0 14px 0;background:linear-gradient(90deg,#0A0E50 0%,#892C75 45%,#E72887 100%);"></div>


Un modelo entrenado resuelve una tarea, pero sus millones de pesos no dicen **cómo** la resuelve. La
**interpretabilidad mecanicista** es la ingeniería inversa de esa solución: identificar **qué variables** representa el
modelo, **dónde** quedan escritas, y **qué componentes** (capas, cabezas) las calculan y combinan, es decir, el
**circuito**. Y, sobre todo, mostrar con **evidencia** que el modelo de verdad usa ese mecanismo.

### ¿Para qué sirve?

- **Explicar y predecir:** si conocemos el algoritmo, podemos predecir qué hará el modelo en casos nuevos.
- **Detectar atajos:** dos modelos con la misma exactitud pueden resolver la tarea de formas distintas, y una de ellas
  puede romperse fuera de la distribución de entrenamiento (lo verá en el desafío final).
- **Confiar y auditar:** saber por qué un modelo responde lo que responde, y no solo cuánto acierta.
- **Intervenir:** si sabemos dónde está una variable, podemos modificarla y cambiar el comportamiento a propósito.

### Ingredientes de un análisis

1. **Un comportamiento bien definido**, con una regla conocida, para saber qué buscar y comparar.
2. **Un modelo que lo resuelva** (casi) siempre: si falla a menudo, no hay un mecanismo limpio que explicar.
3. **Una hipótesis falsable:** qué variables, en qué componentes, y qué predice si la intervenimos.
4. **Herramientas** que aportan **distintos tipos de evidencia** (abajo).
5. **Controles y líneas base:** comparaciones que descartan explicaciones alternativas.

### Tipos de evidencia

| tipo | qué hace la herramienta | qué permite afirmar |
|---|---|---|
| observación | mira activaciones sin tocarlas | qué ocurre junto con el comportamiento |
| plausibilidad | construye el mecanismo a mano | que el mecanismo es *posible* en esta arquitectura |
| consistencia | mide perfiles que la hipótesis exige | que el modelo se comporta *como si* usara el mecanismo |
| causal | interviene una activación y mira la salida | que el modelo *usa* esa activación para responder |
| correlacional | lee variables en las representaciones | que la variable está *escrita* (legible), no que se use |

## La tarea del puntero

**Tokens.** Hay dos tipos:

- **letras simples:** `a` … `h` (8 tokens);
- **marcadores:** una letra con un dígito, `a1` … `h9` (72 tokens). Cada marcador es **un solo token**.

**Secuencias.** 12 tokens de contexto con **4 marcadores** (de letras distintas) en posiciones al azar, y al final una
letra simple: la **consulta**.


<div style="display:flex;align-items:center;background:#EEF1FA;border-left:6px solid #002060;border-radius:8px;padding:12px 18px;margin:8px 0;color:#0A0E50;font-family:Quicksand, 'Segoe UI', Helvetica, Arial, sans-serif;text-align:left;"><div style="flex:1;"><div style="font-weight:700;font-size:1.05em;color:#002060;margin-bottom:6px;">💡 Regla de la tarea</div><div style="line-height:1.55;color:#0A0E50;">Para responder:<ol style="margin:8px 0 2px 0;padding-left:1.4em;"><li>busque el marcador con la letra de la consulta (<code>c</code> → <code>c3</code>);</li><li>desde ese marcador, <b>retroceda tantos tokens como dice su dígito</b> (3);</li><li>responda la letra que está ahí.</li></ol></div></div></div>



En el ejemplo, `c3` está en la posición 7; tres atrás está la posición 4, donde hay una `d`. En símbolos: si la consulta
es la letra $\ell$ y el marcador $\ell d$ está en la posición $m$, la respuesta es $x_{m-d}$. El dígito de un marcador
siempre apunta a una letra simple.

La celda siguiente define el vocabulario y `generar(n, seed)`, que produce secuencias de la tarea. Además de los tokens
devuelve la respuesta, la posición del marcador consultado y la posición de la respuesta: las usaremos para medir y
para intervenir.

````python
#@title Funciones auxiliares: vocabulario y generador { display-mode: "form" }
LETRAS = 'abcdefgh'
S = len(LETRAS)        # letras simples
NDIG = 9               # dígitos de los marcadores: 1 … 9
N_MARC = 4             # marcadores por secuencia
CTX = 12               # tokens de contexto
T = CTX + 1            # + la consulta (última posición)
V = S + S * NDIG       # vocabulario: 8 letras + 72 marcadores


def marcador(letra, d):
    """Token del marcador con esa letra (0 … 7) y ese dígito (1 … 9)."""
    return S + letra * NDIG + (d - 1)


def letra_de(t):  return t if t < S else (t - S) // NDIG
def digito_de(t): return 0 if t < S else (t - S) % NDIG + 1
def nombre(t):    return LETRAS[t] if t < S else f'{LETRAS[letra_de(t)]}{digito_de(t)}'
def texto(x):     return ' '.join(nombre(int(t)) for t in x)


def codificar(s):
    """'a b c3 … c' → lista de tokens."""
    return [LETRAS.index(w) if len(w) == 1 else marcador(LETRAS.index(w[0]), int(w[1])) for w in s.split()]


def generar(n, seed, fijos=False):
    """n secuencias de la tarea. Devuelve:
       X   [n, T]  tokens (la consulta es X[:, -1])
       Y   [n]     respuesta correcta
       MK  [n]     posición del marcador consultado
       OBJ [n]     posición de la respuesta (MK - dígito)
    fijos=True pone los 4 marcadores siempre al final del contexto (lo usamos en el desafío final)."""
    rng = np.random.default_rng(seed)
    partes, total = [], 0
    while total < n:
        m = 2 * n
        x = rng.integers(S, size=(m, CTX))
        if fijos:
            pos = np.tile(np.arange(CTX - N_MARC, CTX), (m, 1))
        else:
            pos = np.sort(np.argsort(rng.random((m, CTX - 1)), 1)[:, :N_MARC] + 1, 1)
        letras = np.argsort(rng.random((m, S)), 1)[:, :N_MARC]          # 4 letras distintas
        es_marc = np.zeros((m, CTX), bool)
        np.put_along_axis(es_marc, pos, True, 1)
        destino = pos[:, :, None] - np.arange(1, NDIG + 1)               # [m, marcador, dígito]
        valido = (destino >= 0) & ~np.take_along_axis(es_marc, destino.clip(0).reshape(m, -1), 1).reshape(destino.shape)
        dig = np.argmax(np.where(valido, rng.random(valido.shape), -1), -1) + 1   # un dígito válido al azar
        np.put_along_axis(x, pos, marcador(letras, dig), 1)
        r = np.arange(m)
        j = rng.integers(N_MARC, size=m)                                  # qué marcador se consulta
        mk = pos[r, j]
        obj = mk - dig[r, j]
        ok = valido.any(-1).all(-1)                                       # descartar si algún marcador no tiene destino
        partes.append((np.concatenate([x, letras[r, j][:, None]], 1)[ok], x[r, obj][ok], mk[ok], obj[ok]))
        total += ok.sum()
    return tuple(np.concatenate(z)[:n] for z in zip(*partes))


X, Y, MK, OBJ = generar(5, seed=1)
for x, y, mk, obj in zip(X, Y, MK, OBJ):
    print(f'{texto(x):42s} →  {nombre(y)}     (marcador en {mk}, respuesta en {obj})')
````

**Dos reglas de referencia.** El *azar* acierta 1 de 8. La **inducción** de la clase respondería lo que siguió a la
última aparición simple de la letra consultada (`… c a … c → a`). ¿Cuánto acierta?

````python
X_test, Y_test, MK_test, OBJ_test = generar(2000, seed=123)


def regla_induccion(x):
    """Lo que siguió a la última aparición simple de la letra consultada (-1 si no aparece)."""
    antes = [i for i in range(CTX - 1) if x[i] == x[-1]]
    return x[antes[-1] + 1] if antes else -1


induccion = np.mean([regla_induccion(x) == y for x, y in zip(X_test, Y_test)])
print(f'azar: {1 / S:.1%}   ·   inducción: {induccion:.1%}')
````

## El modelo y su entrenamiento

El mismo Transformer de la clase: **solo atención**, una cabeza por capa, sin MLP ni LayerNorm, con RoPE (la atención
sabe de *distancias* entre posiciones, no de posiciones absolutas). La pérdida se mide solo en la última posición.

`modelo.run(x)` devuelve los logits y un **caché** con todo lo que necesitamos para interpretar:

| clave | qué es | forma |
|---|---|---|
| `resid0` | embedding | `[n, T, 64]` |
| `atn1`, `atn2` | patrón de atención de la capa 1 y 2 (fila = desde, columna = hacia) | `[n, T, T]` |
| `sal1`, `sal2` | lo que escribe cada capa en el residual | `[n, T, 64]` |
| `resid1`, `resid2` | residual tras la capa 1 y 2 | `[n, T, 64]` |

Con `ganchos={clave: f}` reemplazamos esa activación por `f(activación)`: así intervenimos (paso 5).

````python
#@title Modelo de solo atención, entrenamiento y evaluación { display-mode: "form" }
def rope(x, base=100.0):
    """Rota cada par de coordenadas de q y k según la posición (RoPE)."""
    n, dh = x.shape[-2], x.shape[-1]
    f = base ** (-torch.arange(0, dh, 2, device=x.device).float() / dh)
    a = torch.arange(n, device=x.device).float()[:, None] * f[None]
    c, s = a.cos(), a.sin()
    x1, x2 = x[..., 0::2], x[..., 1::2]
    return torch.stack([x1 * c - x2 * s, x1 * s + x2 * c], -1).flatten(-2)


class SoloAtencion(nn.Module):
    """Transformer de solo atención: una cabeza por capa, sin MLP ni LayerNorm, con RoPE."""

    def __init__(self, capas, d=64, d_cabeza=32):
        super().__init__()
        self.capas, self.dh = capas, d_cabeza
        self.E = nn.Embedding(V, d)
        lineales = lambda a, b: nn.ModuleList(nn.Linear(a, b, bias=False) for _ in range(capas))
        self.Q, self.K, self.Vv = lineales(d, d_cabeza), lineales(d, d_cabeza), lineales(d, d_cabeza)
        self.O = lineales(d_cabeza, d)
        self.U = nn.Linear(d, S, bias=False)            # la respuesta siempre es una letra simple

    def run(self, x, ganchos=None):
        ganchos = ganchos or {}
        igual = lambda t: t
        L = x.shape[1]
        futuro = torch.triu(torch.ones(L, L, dtype=torch.bool, device=x.device), 1)
        cache = {}
        h = ganchos.get('resid0', igual)(self.E(x))
        cache['resid0'] = h
        for l in range(1, self.capas + 1):
            q, k, v = rope(self.Q[l - 1](h)), rope(self.K[l - 1](h)), self.Vv[l - 1](h)
            A = torch.softmax((q @ k.transpose(1, 2) / math.sqrt(self.dh)).masked_fill(futuro, -1e9), -1)
            A = ganchos.get(f'atn{l}', igual)(A)
            cache[f'atn{l}'] = A
            o = ganchos.get(f'sal{l}', igual)(self.O[l - 1](A @ v))
            cache[f'sal{l}'] = o
            h = ganchos.get(f'resid{l}', igual)(h + o)
            cache[f'resid{l}'] = h
        return self.U(h), cache


def correr(modelo, X, ganchos=None):
    """Logits de la última posición y caché, sin gradientes."""
    with torch.no_grad():
        logits, cache = modelo.run(torch.as_tensor(X, device=DEV), ganchos)
    return logits[:, -1], cache


def predecir(modelo, X, ganchos=None):
    return correr(modelo, X, ganchos)[0].argmax(-1).cpu().numpy()


def exactitud(modelo, X, Y, ganchos=None):
    return float(np.mean(predecir(modelo, X, ganchos) == Y))


def entrenar(capas, pasos=5000, lote=256, lr=3e-3, semilla=0, fijos=False):
    torch.manual_seed(semilla)
    modelo = SoloAtencion(capas).to(DEV)
    opt = torch.optim.AdamW(modelo.parameters(), lr, weight_decay=0.01)
    sched = torch.optim.lr_scheduler.OneCycleLR(opt, lr, total_steps=pasos, pct_start=0.05)
    X_val, Y_val = generar(1000, seed=999, fijos=fijos)[:2]
    curva, t0 = [], time.time()
    for paso in range(pasos + 1):
        if paso % 100 == 0:
            curva.append((paso, exactitud(modelo, X_val, Y_val)))
        if paso == pasos:
            break
        Xb, Yb = generar(lote, seed=semilla * 10**6 + paso + 1, fijos=fijos)[:2]
        logits = modelo.run(torch.as_tensor(Xb, device=DEV))[0][:, -1]
        loss = F.cross_entropy(logits, torch.as_tensor(Yb, device=DEV))
        opt.zero_grad()
        loss.backward()
        opt.step()
        sched.step()
    print(f'{capas} capa(s): exactitud {curva[-1][1]:.1%} en {time.time() - t0:.0f}s')
    return modelo.eval(), curva
````

````python
# TODO: Complete el código a continuación

def atencion_media(modelo, X, MK, OBJ):
    """Atención media de cada capa entre las posiciones que importan para la tarea."""
    _, c = correr(modelo, X)
    A1, A2 = c['atn1'].cpu().numpy(), c['atn2'].cpu().numpy()   # A[b, desde, hacia]: un valor por secuencia b
    n, q = len(X), CTX                                           # q: posición de la consulta (la última)
    r = np.arange(n)                                             # índice de cada secuencia: 0 … n-1
    filas, cols = np.nonzero(X[:, :CTX] >= S)                     # cada marcador: secuencia filas[m], posición cols[m]
    d = np.array([digito_de(t) for t in X[filas, cols]])          # dígito de cada marcador
    # MK[b]: posición del marcador consultado en la secuencia b · OBJ[b]: posición de la respuesta

    res = {}

    # EJEMPLO. Capa 1, desde cada marcador (i) hacia i - d. Promedio sobre todos los marcadores.
    res['capa 1 · marcador -> i - d'] = A1[filas, cols, cols - d].mean()

    # TODO 1. Capa 1, desde cada marcador (i) hacia i - 1, el token anterior.
    #         Promedio solo sobre los marcadores con d != 1 (con d = 1, i - 1 ya es el destino correcto).
    res['capa 1 · marcador -> i - 1 (control)'] = None

    # TODO 2. Capa 1, desde la consulta (q) hacia SU marcador (MK). Promedio sobre las n secuencias.
    res['capa 1 · consulta -> su marcador'] = None

    # TODO 3. Lo mismo que el TODO 2, pero en la capa 2.
    res['capa 2 · consulta -> su marcador'] = None

    # TODO 4. Capa 2, desde la consulta (q) hacia la posición de la respuesta (OBJ). Promedio sobre las n secuencias.
    res['capa 2 · consulta -> la respuesta'] = None

    # TODO 5. Capa 2, desde la consulta (q) hacia cada marcador que NO es el consultado (cols != MK[filas]).
    #         Promedio sobre todos esos marcadores: un valor por marcador, no por secuencia.
    res['capa 2 · consulta -> otro marcador (cada uno)'] = None

    # TODO 6. Capa 2, desde la consulta (q): en cada secuencia, SUME la atención a todas las posiciones del contexto
    #         donde está la misma letra simple que la consulta (X[b, j] == X[b, q], j < CTX).
    #         Luego promedie esas sumas sobre las n secuencias.
    res['capa 2 · consulta -> letra simple igual (suma)'] = None
    return res


for k, v in atencion_media(modelo, X_test, MK_test, OBJ_test).items():
    print(f'{k:48s} {v if v is None else round(float(v), 2)}')
````

## Paso 3 · Demostrar a mano: ¿es posible el mecanismo?



Construimos el mecanismo **sin entrenar**, escribiendo las puntuaciones de atención a mano. Si con la hipótesis se
resuelve la tarea, la hipótesis es al menos *posible* en esta arquitectura.

**Capa 1 con RoPE.** RoPE hace que la puntuación entre *i* y *j* dependa de la distancia *i − j* a través de cosenos.
Si la query de un marcador con dígito *d* está rotada en *d·θ*, la puntuación es

$$s^{(1)}_{ij} = \beta_1 \cos\big((i - j - d_i)\,\theta\big),$$

máxima justo en $j = i - d_i$.

**Capa 2.** Solo las letras simples preguntan y solo los marcadores responden. La query de una letra simple es su letra
en *one-hot*, $q_i = e_{\text{letra}(x_i)}$ (cero si $x_i$ es un marcador); la key de un marcador es la letra del
marcador, $k_j = e_{\text{letra}(x_j)}$ (cero si $x_j$ es una letra simple). La puntuación es su producto punto:

$$s^{(2)}_{ij} = \beta_2\,\langle q_i, k_j\rangle =
\begin{cases} \beta_2 & \text{si } x_j \text{ es el marcador de la letra } x_i,\\ 0 & \text{en otro caso,} \end{cases}$$

así que desde la consulta `c` la atención se concentra en `c3`, y la capa 2 copia lo que la capa 1 guardó ahí.

### ¿Qué planos de RoPE usa cada capa?

Caracterizar no es solo ver a dónde mira cada capa: también es ver **con qué** mira. RoPE divide la query y la key en
16 **planos** (pares de coordenadas) y rota cada uno según la posición, con un ángulo fijo por paso $\theta_p$: de
57° por posición en el plano más rápido a menos de 1° en el más lento. En el plano $p$, la puntuación entre $i$ y $j$
depende de la distancia a través de $\cos\big((i - j)\,\theta_p + \varphi\big)$, donde el ángulo $\varphi$ lo fija el
contenido de $q$ y $k$:

- en un plano **rápido**, el coseno cambia mucho de una distancia a la siguiente: sirve para **elegir una distancia**;
- en un plano **lento**, el coseno casi no cambia en 12 posiciones: la puntuación depende casi solo del **contenido**.

Medimos cuánto usa cada capa cada plano: la norma media de $q$ (en las posiciones que preguntan) por la norma media
de $k$ (en el contexto), en ese plano y antes de rotar. Es el mismo análisis de las diapositivas «Frecuencias» y
«Distribución» de la clase.

> **Prediga:** ¿qué planos debería usar la capa 1? ¿Y la capa 2?

````python
# ángulo por posición de cada plano de RoPE (base 100), del más rápido al más lento
ang = np.degrees(100.0 ** (-np.arange(0, modelo.dh, 2) / modelo.dh))
_, c = correr(modelo, X_test)
es_marc = X_test[:, :CTX] >= S
uso = {}
with torch.no_grad():
    for l in (1, 2):
        h = c[f'resid{l - 1}']                                    # entrada de la capa l
        norma = lambda W: W(h).view(*h.shape[:2], -1, 2).norm(dim=-1).cpu().numpy()   # ‖·‖ en cada plano, antes de rotar
        q, k = norma(modelo.Q[l - 1]), norma(modelo.K[l - 1])
        q_preg = q[:, :CTX][es_marc] if l == 1 else q[:, CTX]     # quién pregunta: los marcadores (capa 1), la consulta (capa 2)
        u = q_preg.mean(0) * k[:, :CTX].reshape(-1, q.shape[-1]).mean(0)
        uso[l] = u / u.sum()

fig, axs = plt.subplots(1, 2, figsize=(11, 3.3), sharey=True)
for ax, l, col in zip(axs, (1, 2), (GREEN, BLUE)):
    ax.bar(range(len(ang)), uso[l], color=col)
    ax.set_xticks(range(len(ang)), [f'{a:.0f}°' if a >= 10 else f'{a:.1f}°' for a in ang], rotation=60, fontsize=8)
    ax.set_title(f'capa {l}')
    ax.set_xlabel('plano de RoPE (ángulo por posición)')
axs[0].set_ylabel('fracción del uso (‖q‖·‖k‖)')
plt.tight_layout()
plt.show()
````

La celda siguiente define las herramientas de intervención y el donante.

````python
def con_parche(sitio, pos, valores):
    """Ganchos que reemplazan la activación `sitio` en la posición pos[b] de cada secuencia b por valores[b]."""
    pos = torch.as_tensor(pos, device=DEV)

    def f(t):
        t = t.clone()
        t[torch.arange(len(t), device=DEV), pos] = valores
        return t
    return {sitio: f}


def mismo_esqueleto(X, seed):
    """Donante: mismas posiciones, marcadores y consulta; las letras simples se sortean de nuevo."""
    rng = np.random.default_rng(seed)
    D = X.copy()
    simples = D[:, :CTX] < S
    D[:, :CTX][simples] = rng.integers(S, size=simples.sum())
    return D


def resumir(nombre_exp, obtenido, prediccion, receptor):
    """Fracción de casos en que el modelo responde lo predicho, lo del receptor sin intervenir, u otra cosa."""
    pred_es_rec = prediccion == receptor
    fr = {'predicho': np.mean(obtenido == prediccion),
          'receptor': np.mean((obtenido == receptor) & ~pred_es_rec),
          'otro': np.mean((obtenido != prediccion) & (obtenido != receptor))}
    print(f'{nombre_exp:42s}', '   '.join(f'{k} {v:.0%}' for k, v in fr.items()))
    return fr


def elegir(eleccion, opciones):
    """Arreglo de la opción elegida: acepta el texto completo del menú o solo su letra ('a', 'b', …)."""
    letra = eleccion.strip()[:1].lower()
    assert letra in opciones, f'Elija una opción válida ({", ".join(opciones)}), no {eleccion!r}'
    return opciones[letra]


R = np.arange(len(X_test))
X_don = mismo_esqueleto(X_test, seed=7)
_, cache_don = correr(modelo, X_don)
Y_don = X_don[R, OBJ_test]                              # lo que respondería el donante
distinto = Y_don != Y_test                              # nos quedamos con los casos donde el desenlace se distingue
````

## Paso 6 · Descomponer: ¿qué escribe la capa 1 en los marcadores?





Si la hipótesis es correcta, la salida de la capa 1 (`sal1`) en un marcador contiene **la letra que está d atrás**.
Sin usar etiquetas, proyectamos `sal1` de todos los marcadores sobre sus dos primeras componentes principales; después
coloreamos los puntos de tres formas: por la letra guardada, por la letra del marcador y por su dígito.

`pca_2d` centra las activaciones y las proyecta sobre las dos direcciones de mayor varianza (vía SVD). **Ninguna
etiqueta** entra en el cálculo: los colores se agregan después.

````python
def pca_2d(H):
    """Proyección de las filas de H sobre las dos primeras componentes principales."""
    Hc = H - H.mean(0)
    U, s, Vt = np.linalg.svd(Hc, full_matrices=False)
    return Hc @ Vt[:2].T
````

````python
X_pca = generar(1500, seed=21)[0]
_, c = correr(modelo, X_pca)
b_m, i_m = np.nonzero(X_pca[:, :CTX] >= S)                 # todos los marcadores
d_m = np.array([digito_de(t) for t in X_pca[b_m, i_m]])
etiquetas = {
    'letra guardada (en i - d)': X_pca[b_m, i_m - d_m],
    'letra del marcador': np.array([letra_de(t) for t in X_pca[b_m, i_m]]),
    'dígito': d_m,
}
Z = pca_2d(c['sal1'][b_m, i_m].cpu().numpy())

fig, axs = plt.subplots(1, 3, figsize=(15, 4.4))
for ax, (titulo, y) in zip(axs, etiquetas.items()):
    if titulo == 'dígito':
        sc = ax.scatter(Z[:, 0], Z[:, 1], c=y, cmap='cenia_degradado', s=5)
    else:
        ax.scatter(Z[:, 0], Z[:, 1], color=plt.cm.tab10(y), s=5)       # letra k → color tab10(k), igual que la leyenda
    ax.set_title(f'sal1 en los marcadores · color = {titulo}', fontsize=10)
    ax.set_xticks([])
    ax.set_yticks([])
    if titulo == 'dígito':
        fig.colorbar(sc, ax=ax)
    else:
        ax.legend(handles=[plt.Line2D([], [], marker='o', ls='', color=plt.cm.tab10(k), label=LETRAS[k]) for k in range(S)],
                  fontsize=8, loc='center left', bbox_to_anchor=(1, 0.5))
plt.tight_layout()
plt.show()
````

## Paso 7 · Confirmar: probes lineales



Un *probe* es un clasificador lineal que intenta leer una variable desde una activación. Si la hipótesis es correcta:

- en los **marcadores**, la letra guardada no se lee del embedding (`resid0`, ≈ azar = 12.5%) pero sí tras la capa 1
  (`resid1`);
- en la **consulta**, la respuesta solo se lee tras la capa 2 (`resid2`).

`probe` entrena una regresión logística en el 75% de las activaciones y mide su exactitud en el 25% restante.

````python
def probe(H, y, semilla=0):
    """Regresión logística en 75% de las filas; devuelve la exactitud en el 25% restante."""
    idx = np.random.default_rng(semilla).permutation(len(y))
    corte = int(0.75 * len(y))
    tr, te = idx[:corte], idx[corte:]
    clf = LogisticRegression(max_iter=2000).fit(H[tr], y[tr])
    return clf.score(H[te], y[te])
````

````python
X_pr, Y_pr, MK_pr, OBJ_pr = generar(1500, seed=31)
_, c = correr(modelo, X_pr)
b_m, i_m = np.nonzero(X_pr[:, :CTX] >= S)
d_m = np.array([digito_de(t) for t in X_pr[b_m, i_m]])
sitios = ['resid0', 'resid1', 'resid2']

tabla = {}
for var, y in [('marcador · letra guardada', X_pr[b_m, i_m - d_m]),
               ('marcador · su letra', np.array([letra_de(t) for t in X_pr[b_m, i_m]])),
               ('marcador · dígito', d_m)]:
    tabla[var] = [probe(c[s][b_m, i_m].cpu().numpy(), y) for s in sitios]
tabla['consulta · respuesta'] = [probe(c[s][:, CTX].cpu().numpy(), Y_pr) for s in sitios]

plt.figure(figsize=(7, 3.4))
plt.imshow(np.array(list(tabla.values())), cmap='cenia', vmin=0, vmax=1.4, aspect='auto')
plt.grid(False)
for f, fila in enumerate(tabla.values()):
    for s, v in enumerate(fila):
        plt.text(s, f, f'{v:.0%}', ha='center', va='center', color='white' if v > 0.6 else 'black')
plt.xticks(range(3), ['resid0\nembedding', 'resid1\ntras capa 1', 'resid2\ntras capa 2'])
plt.title(f'exactitud del probe (azar = {1 / S:.1%} para letras)', fontsize=10)
plt.yticks(range(len(tabla)), list(tabla))
plt.show()
````
