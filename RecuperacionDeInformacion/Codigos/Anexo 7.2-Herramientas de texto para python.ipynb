## Stopwords

NLTK viene con listas de stop words para distintos idiomas.

````python
import nltk
from nltk.corpus import stopwords

# descargar una lista de palabras. La lista se guarda en una carpeta "nltk_data/corpora/" en el home de usuario.
nltk.download("stopwords")

stoplist = sorted(stopwords.words("spanish"))
print()
print("NLTK define {} stop words para el español:".format(len(stoplist)))
print()

# imprimir las palabras con un ancho maximo
linea = " ".join(stoplist)
ancho_maximo = 120
while True:
    if len(linea) <= ancho_maximo:
        print(linea)
        break
    position = linea.rfind(" ", 0, ancho_maximo)
    if position < 0:
        print(linea)
        break
    print(linea[0:position])
    linea = linea[position + 1 :]
````

## WordNet

Es una base de datos de palabras que se puede descargar desde: https://wordnet.princeton.edu/download/current-version

Para cada palabra está su tipo (sustantivo, verbo, etc.), definición (como un diccionario), sinónimos y relaciones con otras palabras.

Al grupo de palabras con un mismo significado se le llama "synset".

Se puede ver la documentación de NLTK en: https://www.nltk.org/howto/wordnet.html

Junto con este notebook está la base de datos de WordNet:

  * adjetivos `WordNet-3.1_data.adjectives`
  * adverbios `WordNet-3.1_data.adverbs`
  * sustantivos `WordNet-3.1_data.nouns`
  * verbos `WordNet-3.1_data.verbs`.

Cada archivo tiene un synset por fila. 

La descripción del formato está en: https://wordnet.princeton.edu/documentation/wndb5wn

El formato de cada fila es (esencialmente): 
  1. id del synset
  2. las palabras del synset
  3. lista de synset relacionados (usualmente el synset padre o generalización)
  4. definición de diccionario


````python
import nltk
from nltk.corpus import wordnet

nltk.download("wordnet")

lista_synsets = wordnet.synsets("lion")

for syn in lista_synsets:
    print()
    print(syn)
    print("  -definición: {}".format(syn.definition()))
    print("  -palabras: {}".format(syn.lemma_names()))
    print("  -padres: {}".format(syn.hypernyms()))
    print("  -hijos: {}".format(syn.hyponyms()))
````

## Similitud entre conceptos (synsets)


````python
synset1 = wordnet.synset("cat.n.01")
synset2 = wordnet.synset("dog.n.01")

print(synset1.path_similarity(synset2))
````
